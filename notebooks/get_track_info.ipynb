```python
import os
import time
import pandas as pd
```

```python
%%time
from ytmusicapi import YTMusic
yt = YTMusic('../headers_auth.json')

BACKUP_DIR  = '../playlists'
NEW_TRACKS_TSV = os.path.join(BACKUP_DIR, '_tracks.tsv') 
CURRENT_TRACKS_TSV = os.path.join(BACKUP_DIR, '_tracks_orig.tsv') 
```

```python
unique_tracks = pd.read_csv(CURRENT_TRACKS_TSV, sep='\t', index_col=0)
tracks_w_meta = pd.read_csv(NEW_TRACKS_TSV, sep='\t', index_col=0)
# tracks_w_meta_sub = tracks_w_meta.sample(len(tracks_w_meta) - 5)
print('%d unique tracks, and %d with metadata' % (len(unique_tracks), len(tracks_w_meta)))
```

```python
def get_track_info(row):
    copy_cols=['keywords', 'averageRating', 'viewCount', 'release']
    if type(row['albumId']) == str and 'privately_owned' not in row['albumId']:
        try:
            song = yt.get_song(row.name)
            for col in copy_cols:
                if col in song:
                    row[col] = song[col]
        except Exception as e:
            print('Failed to get track info: %s\n%s' % (e, row))
    return row
```

```python
%%time
# tracks_w_meta = unique_tracks.apply(get_track_info, axis=1)

i=0
new_rows = []
for name, row in unique_tracks.iterrows():
    i+=1
    print('(%d/%d): %s - %s - %s' % (i, len(unique_tracks), row['artist'], row['album'], row['title']))
    new_rows.append(get_track_info(row))

tracks_w_meta = pd.DataFrame(new_rows)
tracks_w_meta.to_csv(NEW_TRACKS_TSV, sep='\t', header=True)
assert len(unique_tracks) == len(tracks_w_meta)
```
