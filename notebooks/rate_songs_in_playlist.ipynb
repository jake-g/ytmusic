```python
import os
import time
import pandas as pd
from ytmusicapi import YTMusic
```

```python
%%time
yt = YTMusic('../headers_auth.json')

# Get a set of videoIds for thumbs up and thumbs down
def get_playlist_track_ids(playlist_id, print_stuff=False):
    playlist_meta = yt.get_playlist(playlist_id, limit=10000)
    track_ids = set()
    for i, track in enumerate(playlist_meta['tracks']):
        if print_stuff:
            try:
                print('(%d/%d) %s - %s - %s' % (
                    i+1, len(playlist_meta['tracks']), track['artists'][0]['name'], track['album']['name'], track['title']))
            except Exception as e:
                print(e)
        track_ids.add(track['videoId'])
    return track_ids
```

```python
playlists = pd.DataFrame(yt.get_library_playlists(limit=500))
```

```python
force_like_playlists = set(["beats", "beats wonky LA '10 scene", "electronic Analog Grooves like", "electronic new indie beats", "electronic top", "future beats", "hiphop", "indie loose like", "jazz", "my balls your chill", "nudisco future", "Nu Disco Boogie like", "psych rock modern", "reggae classic", "rock modern chill", "soul funk", "x_r.treemusic like", "y_1950s_thumbs_up", "y_1960s_thumbs_up", "y_1970s_thumbs_up", "y_1980s_thumbs_up", "y_1990s_thumbs_up", "y_2000s_thumbs_up", "y_2005s_thumbs_up", "y_2010_thumbs_up", "y_2011_thumbs_up", "y_2012_thumbs_up", "y_2013_thumbs_up", "y_2014_thumbs_up", "y_2015_thumbs_up", "y_2016_thumbs_up"])
playlist_ids = set()
thumbs_up_ids = set()

# Get track ids from above playlists that are not already liked
for i, row in playlists.iterrows():
    if row.title in force_like_playlists:
        playlist_ids.add(row.playlistId)
        playlist_meta = yt.get_playlist(row.playlistId, limit=10000)
        for i, track in enumerate(playlist_meta['tracks']):
            if track['likeStatus'] != 'LIKE':
                thumbs_up_ids.add(track['videoId'])
        print(f'Finished retrieving tracks from playlist: {row.title}')
print(f'Force like {len(thumbs_up_ids)} tracks from {len(playlist_ids)} playlists')
```

```python
%%time
# Rate the thumbs up/down tracks on ytmusic using api
SLEEP_TIME=0.2

rating='LIKE'
for i, vId in enumerate(thumbs_up_ids):
    print('(%d/%d) %s' % (i+1, len(thumbs_up_ids), vId))
    yt.rate_song(vId, rating=rating)
    time.sleep(SLEEP_TIME) 
```

```python
rating='DISLIKE'
for i, vId in enumerate(thumbs_down_ids):
    print('(%d/%d) %s' % (i+1, len(thumbs_down_ids), vId))
    yt.rate_song(vId, rating=rating)
    time.sleep(SLEEP_TIME) 
```
