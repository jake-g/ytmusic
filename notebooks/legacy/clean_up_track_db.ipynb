# Remove tracks from db with missing info (let them repopulate next run)

```python
yt_track_db_tsv = '/Users/jk/projects/music-library/ytmusic/playlists/_tracks_db.tsv'
df = pd.read_csv(yt_track_db_tsv, sep='\t', index_col=0)
```

```python
rm_ids = (
    frozenset(df.loc[df['albumArtist'].isna()].index) |
    frozenset(df.loc[df['artistId'].isna()].index) |
    frozenset(df.loc[df['albumId'].isna()].index) |
    frozenset(['#NAME?']) # some how this is duplicated alot
)
print('df length', len(df), 'num rm ids', len(rm_ids))
df = df.loc[~df.index.isin(rm_ids)]
print('cleaned df length', len(df))
df.to_csv(yt_track_db_tsv, sep='\t', index=True, header=True)
```
