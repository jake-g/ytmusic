```python
import os
import time
import pandas as pd
```

```python
%%time
from ytmusicapi import YTMusic
yt = YTMusic('../headers_auth.json')
```

```python
%%time
def get_track_info(row, ):
    copy_cols=['keywords', 'averageRating', 'viewCount', 'release']

    if type(row['albumId']) == str and 'privately_owned' not in row['albumId']:
        try:
            song = yt.get_song(row.name)
            for col in copy_cols:
                if col in song:
                    row[col] = song[col]
        except Exception as e:
            print('Failed to get track info: %s\n%s' % (e, row))
    return row

unique_tracks = pd.read_csv('./_tracks.tsv', sep='\t', index_col=0)
tracks_w_meta = unique_tracks.apply(get_track_info, axis=1)

tracks_w_meta
```

```python
_tracks_w_meta = _tracks_w_info

BACKUP_DIR  = './'
TRACKS_TSV = '_tracks_w_meta.tsv'
_tracks_w_meta.to_csv(os.path.join(BACKUP_DIR, TRACKS_TSV), sep='\t', header=True)
```
