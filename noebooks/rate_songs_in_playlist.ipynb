```python
import os
import time
import pandas as pd
from ytmusicapi import YTMusic
```

```python
%%time
# Get Playlists
PLAYLIST_LIMIT=500
yt = YTMusic('headers_auth.json')
playlists = pd.DataFrame(yt.get_library_playlists(limit=PLAYLIST_LIMIT))
```

```python
%%time
# Create set of thumbs up/down tracks from filterwed playlists
thumbs_up_ids = set()
thumbs_down_ids = set()
PLAYLIST_SONG_LIMIT=10000

# Get a set of videoIds for thumbs up and thumbs down
def get_playlist_track_ids(playlist_id):
    playlist_meta = yt.get_playlist(playlist_id, limit=PLAYLIST_SONG_LIMIT)
    track_ids = set()
    for i, track in enumerate(playlist_meta['tracks']):
        try:
            print('(%d/%d) %s - %s - %s' % (
                i+1, len(playlist_meta['tracks']), track['artists'][0]['name'], track['album']['name'], track['title']))
        except Exception as e:
            print(e)
        track_ids.add(track['videoId'])
    return track_ids
```

```python
%%time
# Rate the thumbs up/down tracks on ytmusic using api
SLEEP_TIME=0.2
rating='DISLIKE'
for i, vId in enumerate(thumbs_down_ids):
    print('(%d/%d) %s' % (i+1, len(thumbs_down_ids), vId))
    yt.rate_song(vId, rating=rating)
    time.sleep(SLEEP_TIME) 

rating='LIKE'
for i, vId in enumerate(thumbs_up_ids):
    print('(%d/%d) %s' % (i+1, len(thumbs_up_ids), vId))
    yt.rate_song(vId, rating=rating)
    time.sleep(SLEEP_TIME) 
```
